# Model Training and Evaluation

This notebook runs a stratified train/test split, five-fold GridSearchCV for Logistic Regression and Random Forest, one-hot encoding for nominal fields, and separate held-out test reporting. Training runs log parameters, metrics, models, EDA plots, and evaluation plots to MLflow.

```python
from heart_disease_mlop.data import load_data, prepare_features
from heart_disease_mlop.modeling import evaluate_models, train_and_track

df = load_data('../data/heart.csv')
X, y = prepare_features(df)
print(f'Prepared features: {X.shape[1]} columns, {len(X)} complete rows')

metrics = evaluate_models('../data/heart.csv')
metrics
```

```python
for model_name, model_metrics in metrics.items():
    print(f'--- {model_name} ---')
    print('Best parameters:', model_metrics['best_params'])
    print('Five-fold CV mean/std:', model_metrics['cross_validation'])
    print('Held-out test:', model_metrics['held_out_test'])
```

## Tuning and best model selection
The parameter grids are searched using five-fold stratified cross-validation on the development partition only. The estimator with the highest mean CV ROC-AUC is selected and refit on the full development partition. The held-out 20% is reserved for final metrics and is never used for tuning.
