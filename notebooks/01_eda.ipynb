# Heart Disease EDA

This notebook loads the cleaned UCI heart disease dataset and performs exploratory analysis.

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from pathlib import Path

df = pd.read_csv('../data/heart_cleaned.csv')
df.head()
```

```python
print(df.shape)
print(df['target'].value_counts())
df.describe().round(2)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='target')
plt.title('Heart Disease Class Balance')
plt.tight_layout()
plt.savefig('../screenshots/class_distribution.png', dpi=150)
plt.close()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.tight_layout()
plt.savefig('../screenshots/age_distribution.png', dpi=150)
plt.close()
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(11, 9))
sns.heatmap(corr, cmap='coolwarm', annot=False)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('../screenshots/correlation_heatmap.png', dpi=150)
plt.close()
```
